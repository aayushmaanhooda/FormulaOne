```python
import fastf1
import os
```

```python
# Create the cache folder if it doesn't exist
if not os.path.exists('f1_cache'):
    os.makedirs('f1_cache')

# Enable caching (saves data locally so you don't download every time)
fastf1.Cache.enable_cache('f1_cache')

# Load a race session
session = fastf1.get_session(2024, 'Monaco', 'R')  # 2024 Monaco Grand Prix Race
session.load()

# Pick a driver's fastest lap
lec_lap = session.laps.pick_driver('LEC').pick_fastest()  # Leclerc's fastest lap

# Get telemetry for that lap
telemetry = lec_lap.get_telemetry()

print(telemetry)
```
