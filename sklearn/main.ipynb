```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
X,y = load_breast_cancer(return_X_y=True)
```

```python
X_train , X_test, y_train, y_test = train_test_split(X,y , test_size=0.2)
```

```python
scaler = StandardScaler()
# use minmax scalr if you want it in range of 0-1


X_train_scaled = scaler.fit_transform(X_train)
# we dont do fit again coz we want to use same mean and variance not different ones
X_test_scaled = scaler.transform(X_test)

print("before transform", X_train)
print("after transform", X_train_scaled)
```

```python
knn  = KNeighborsClassifier()

knn.fit(X_train_scaled, y_train)
```

```python
print(knn.score(X_test_scaled, y_test))
```

```python
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = fetch_openml("car", as_frame=True).frame
```

```python
data
```

```python
columns_to_encode = ["lug_boot", "safety"]

encoder = OrdinalEncoder(
    categories=[
        ["small", "med", "big"],
        ["low", "med", "high"],
    ]
)
data[columns_to_encode] = encoder.fit_transform(data[columns_to_encode])
data
```

```python
encoder.inverse_transform(data[columns_to_encode])
```
