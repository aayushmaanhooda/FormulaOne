```python
import fastf1 
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import os


if not os.path.exists('f1_cache'):
    os.makedirs('f1_cache')

# Now enable cache
fastf1.Cache.enable_cache('f1_cache')
```

```python
# Step 1: Get the 2024 schedule
schedule = fastf1.get_event_schedule(2024)

# step 2 : get all events names
event_names = schedule["EventName"].tolist()[1:]
```

```python
# Step 3: Collect qualifying and race data for each event


qualifying_data = pd.DataFrame()
race_data = pd.DataFrame()

for event in event_names:
    try:
        # qualifying data
        qualifying_session = fastf1.get_session(2024, event, "Q")
        qualifying_session.load()

        #  take copy or else it will think its a view and not concat
        quali_result = qualifying_session.results[["Abbreviation", "TeamName", "Position", "Q1"]].copy()
        quali_result["Event"] = event

        qualifying_data = pd.concat([qualifying_data, quali_result], ignore_index=True)
        print(f"✓ qualifying- {event} — Total rows: {len(qualifying_data)}")

        # race session
        race_session = fastf1.get_session(2024, event, "Race")
        race_session.load()
        race_result = race_session.results[["Abbreviation", "TeamName", "Position", "Status", "FullName"]]
        race_data  = pd.concat([race_data, race_result], ignore_index=True)
        print(f"✓ race - {event} — Total rows: {len(race_data)}")

       
    
    except Exception as e:
        print(f"✗ Failed: {event} - {e}")
        continue
```

```python
qualifying_data['Q1_seconds'] = qualifying_data['Q1_seconds'].fillna(qualifying_data['Q1_seconds'].max() + 1)  # 1 seconds slower than worst
qualifying_data
```
